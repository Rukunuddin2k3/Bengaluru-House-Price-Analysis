```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(subset=["location"], inplace=True)
```

```python
df["size"]=df["size"].str.strip("BHK")
df["size"]=df["size"].str.strip("Bedroom")
df["size"]=df["size"].str.strip("Room")
df["size"].unique()
df["size"]=df["size"].astype("float")
```

```python
df["size"]=df["size"].fillna(df["size"].median())
```

```python
df["society"].fillna("Not Available", inplace=True)
```

```python
df.info()
```

```python

df["society"].count()

df["society"].fillna("unknown")
df["society"].isnull().sum()
```

```python
df["bath"] = df["bath"].fillna(df["bath"].median())
```

```python
df["balcony"] = df["balcony"].fillna(0)
```

```python
df.info()
```

```python
df["total_sqft"].unique()
def convert_sqft(x):
    tokens = x.split('-')
    if len(tokens) == 2:
        return (float(tokens[0]) + float(tokens[1])) / 2
    try:
        return float(x)
    except:
        return None

df["total_sqft"] = df["total_sqft"].apply(convert_sqft)
```

```python
df = df.dropna(subset=["total_sqft"])
```

```python
df["price_per_sqft"] = (df["price"] * 100000) / df["total_sqft"]
df.rename(columns={"size": "bhk"}, inplace=True)
df
```

```python
df.to_csv("../data/cleaned_house_prices.csv", index=False)
```
